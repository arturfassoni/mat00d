## Exercícios

Exercícios retirados da 9a Edição do livro _Equações Diferenciais e Problemas de Valores de Contorno_ de Boyce e DiPrima.

### Introdução

**Exercício 1.** [Seção 1.1, Exerc. 1-4] Em cada um dos itens abaixo, desenhe um campo de direçöes para a equação diferencial dada. Baseado no campo de direções, determine o comportamento de $y$ quando $t \rightarrow \infty$. Se esse comportamento depender do valor inicial de $y$ em $t=0$, descreva essa dependência.

a. $y'=3-2 y$

b. $y'=2 y-3$

c. $y'=-1-2 y$

d. $y'=1+2 y$


**Exercício 2.** [Seção 1.3, Exerc. 5-7] Em cada um dos itens abaixo, verifique que cada função dada é solução da equação diferencial.

a. $\quad y^{\prime \prime}-y=0 ; \quad y_1(t)=e^t, \quad y_2(t)=\cosh t$

b. $\quad y^{\prime \prime}+2 y^{\prime}-3 y=0 ; \quad y_1(t)=e^{-3 t}, \quad y_2(t)=e^t$

c. $\quad t y^{\prime}-y=t^2 ; \quad y=3 t+t^2$

**Exercício 3.** [Seção 1.3, Exerc. 1-4] Em cada um dos itens abaixo, determine a ordem da equação diferencial dada e diga se ela é linear ou não linear.

a. $t^2 \frac{d^2 y}{d t^2}+t \frac{d y}{d t}+2 y=\operatorname{sen} t$

b. $\left(1+y^2\right) \frac{d^2 y}{d t^2}+t \frac{d y}{d t}+y=e^t$

c. $\frac{d^4 y}{d t^4}+\frac{d^3 y}{d t^3}+\frac{d^2 y}{d t^2}+\frac{d y}{d t}+y=1$

d. $\frac{d^2 y}{d t^2}+\operatorname{sen}(t+y)=\operatorname{sen} t$
