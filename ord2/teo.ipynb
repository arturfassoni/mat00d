# Teoria Geral: Existência, Unicidade e Solução Geral

Nas últimas três aulas, resolvemos $ay''+by'+cy=0$ nos três casos possíveis do discriminante.

Guiados pela intuição, chutamos $y=e^{rt}$, encontramos os casos em que isso falha (raiz repetida) ou 'quebra' (raízes complexas), e consertamos cada um.

Funcionou muito bem, mas fica uma pergunta em aberto: como sabemos que não existe nenhuma outra solução escondida, fora da família que encontramos em cada caso?

Esta aula fecha essa lacuna com um teorema geral. É um resultado que vale até para equações lineares com coeficientes variáveis, mas, como neste curso só trabalhamos com coeficientes constantes, vamos usá-lo apenas para confirmar que os três casos já resolvidos realmente esgotam todas as soluções possíveis.

## Um Teorema Geral

**Teorema.** Considere o problema de valor inicial

$$
ay''+by'+cy=0, \qquad y(t_0)=y_0,\ \ y'(t_0)=y_0',
$$

com $a,b,c$ constantes reais, $a\neq0$. Esse problema sempre tem **uma única solução**, definida para todo $t$ real.

Além disso, se $y_1(t)$ e $y_2(t)$ são duas soluções da equação $ay''+by'+cy=0$ tais que o **Wronskiano**

$$
W(y_1,y_2)(t) = y_1(t)y_2'(t) - y_2(t)y_1'(t)
$$

não se anula em *algum* ponto $t$, então **toda solução** da equação pode ser escrita como combinação linear destas duas soluções,

$$
y(t) = c_1y_1(t) + c_2y_2(t)
$$

para alguma escolha das constantes $c_1,c_2$. Nesse caso, dizemos que $\{y_1,y_2\}$ é um **conjunto fundamental de soluções**.

### Por que o Wronskiano?

Não vamos provar esse teorema aqui, mas vale entender **por que** a condição $W\neq0$ faz tanto sentido.

Lembra do que fizemos, exemplo após exemplo: dadas duas soluções $y_1,y_2$, para achar $c_1,c_2$ que satisfazem $y(t_0)=y_0$ e $y'(t_0)=y_0'$, montamos e resolvemos o sistema linear

$$
\begin{cases}
c_1y_1(t_0) + c_2y_2(t_0) = y_0\\[2pt]
c_1y_1'(t_0) + c_2y_2'(t_0) = y_0'
\end{cases}
$$

Da álgebra linear, esse sistema $2\times2$ tem solução única para **quaisquer** $y_0,y_0'$ exatamente quando o determinante da matriz de coeficientes é diferente de zero:

$$
\det\begin{pmatrix} y_1(t_0) & y_2(t_0) \\ y_1'(t_0) & y_2'(t_0)\end{pmatrix} = y_1(t_0)y_2'(t_0)-y_2(t_0)y_1'(t_0) = W(y_1,y_2)(t_0).
$$

Ou seja, $W\neq0$ é exatamente a condição para o sistema que já resolvíamos, na marra, em todos os exemplos anteriores. O Wronskiano só dá um nome (e uma checagem rápida, sem precisar de nenhuma condição inicial específica) para algo que já fazíamos manualmente toda vez.

## Verificando os três casos

Vamos conferir que, nos três casos já resolvidos, o par de soluções fundamentais $y_1,y_2$ encontrado realmente tem Wronskiano diferente de zero, confirmando que encontramos, de fato, *todas* as soluções em cada caso.

```python
import sympy as sp

t, r, r1, r2, lam, mu = sp.symbols('t r r1 r2 lambda mu', real=True)

def wronskiano(y1, y2, var):
    return sp.simplify(sp.factor(y1*sp.diff(y2, var) - y2*sp.diff(y1, var)))

# Caso 1: raízes reais distintas
y1a, y2a = sp.exp(r1*t), sp.exp(r2*t)
print(r"Raízes reais distintas:  W =", wronskiano(y1a, y2a, t))

# Caso 2: raiz repetida
y1b, y2b = sp.exp(r*t), t*sp.exp(r*t)
print(r"Raiz repetida:           W =", wronskiano(y1b, y2b, t))

# Caso 3: raízes complexas
y1c, y2c = sp.exp(lam*t)*sp.cos(mu*t), sp.exp(lam*t)*sp.sin(mu*t)
print(r"Raízes complexas:        W =", wronskiano(y1c, y2c, t))
```

* **Raízes reais distintas** ($r_1\neq r_2$): $W=(r_2-r_1)e^{(r_1+r_2)t}$. Como $r_1\neq r_2$ e a exponencial nunca é zero, $W\neq0$ sempre.
* **Raiz repetida**: $W=e^{2rt}$, que nunca é zero.
* **Raízes complexas** ($\mu\neq0$, já que $\mu$ é a parte imaginária de uma raiz genuinamente complexa): $W=\mu\,e^{2\lambda t}$, que também nunca é zero.

Assim, o teorema confirma que não deixamos nada de fora: nos três casos, o par de soluções que encontramos por intuição já era, o tempo todo, um conjunto fundamental de soluções.

## Resumo dos três casos

| $\Delta=b^2-4ac$ | Raízes | Solução geral |
|---|---|---|
| $\Delta>0$ | $r_1\neq r_2$ reais | $y=c_1e^{r_1t}+c_2e^{r_2t}$ |
| $\Delta=0$ | $r$ real, repetida | $y=(c_1+c_2t)e^{rt}$ |
| $\Delta<0$ | $r=\lambda\pm i\mu$ | $y=e^{\lambda t}(c_1\cos\mu t+c_2\,\text{sen}\,\mu t)$ |

Essas três fórmulas resolvem *qualquer* equação $ay''+by'+cy=0$ com coeficientes constantes.

### Três exemplos-relâmpago para fixar as ideias

Apresentamos abaixo um exemplo de cada caso, só para fixar a relação entre a equação e a solução geral (sem resolver PVI):

**$\Delta>0$:** $y''-y=0 \ \Rightarrow\ r=\pm1 \ \Rightarrow\ y=c_1e^{t}+c_2e^{-t}$.

**$\Delta=0$:** $y''-6y'+9y=0 \ \Rightarrow\ r=3 \ \Rightarrow\ y=(c_1+c_2t)e^{3t}$.

**$\Delta<0$:** $y''-2y'+5y=0 \ \Rightarrow\ r=1\pm2i \ \Rightarrow\ y=e^{t}(c_1\cos2t+c_2\,\text{sen}\,2t)$.
