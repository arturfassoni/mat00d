## Exercícios

Exercícios retirados da 11a Edição do livro _Equações Diferenciais e Problemas de Valores de Contorno_ de Boyce e DiPrima.

### Equações Homogêneas com Coeficientes Constantes

**Exercício 1.** [Seção 3.1, Exerc. 5] Encontre a solução geral da equação diferencial dada:

$$
4y^{\prime\prime}-9y=0.
$$

**Exercício 2.** [Seção 3.1, Exerc. 8,9] Em cada um dos itens abaixo, encontre a solução do problema de valor inicial dado. Esboce o gráfico da solução e descreva seu comportamento quando $t$ aumenta.

a. $y^{\prime\prime}+4y^{\prime}+3y=0, \quad y(0)=2,\ y^{\prime}(0)=-1$

b. $y^{\prime\prime}+3y^{\prime}=0, \quad y(0)=-2,\ y^{\prime}(0)=3$

**Exercício 3.** [Seção 3.1, Exerc. 15] Encontre a solução do problema de valor inicial

$$
2y^{\prime\prime}-3y^{\prime}+y=0, \qquad y(0)=2,\ y^{\prime}(0)=\frac12.
$$

Depois determine o valor máximo da solução e encontre, também, o ponto em que a solução se anula.

**Exercício 4.** [Seção 3.1, Exerc. 16] Resolva o problema de valor inicial

$$
y^{\prime\prime}-y^{\prime}-2y=0, \qquad y(0)=\alpha,\ y^{\prime}(0)=2.
$$

Depois encontre $\alpha$ de modo que a solução tenda a zero quando $t\to\infty$.

### Soluções de Equações Lineares Homogêneas; o Wronskiano

**Exercício 5.** [Seção 3.2, Exerc. 11] Verifique que $y_1(t)=1$ e $y_2(t)=t^{1/2}$ são soluções da equação diferencial $yy^{\prime\prime}+(y^{\prime})^2=0$ para $t>0$. Então, mostre que $y=c_1+c_2t^{1/2}$ não é, em geral, solução desta equação. Explique por que este resultado não contradiz o Teorema visto na seção Teoria Geral.

**Exercício 6.** [Seção 3.2, Exerc. 14] Se o wronskiano $W$ de $f$ e $g$ for $3e^{4t}$ e se $f(t)=e^{2t}$, encontre $g(t)$.

**Exercício 7.** [Seção 3.2, Exerc. 17] Encontre um conjunto fundamental de soluções para a equação diferencial e o ponto inicial dados:

$$
y^{\prime\prime}+y^{\prime}-2y=0, \qquad t_0=0.
$$

**Exercício 8.** [Seção 3.2, Exerc. 21] Verifique que as funções $y_1$ e $y_2$ são soluções da equação diferencial dada. Elas formam um conjunto fundamental de soluções?

$$
x^2y^{\prime\prime}-x(x+2)y^{\prime}+(x+2)y=0, \quad x>0; \qquad y_1(x)=x,\ \ y_2(x)=xe^x.
$$

**Exercícios 9,10,11.** _Retirados_.

### Raízes Complexas

**Exercício 12.** [Seção 3.3, Exerc. 8] Encontre a solução geral da equação diferencial dada:

$$
y^{\prime\prime}+6y^{\prime}+13y=0.
$$

**Exercício 13.** [Seção 3.3, Exerc. 12,13,14] Em cada um dos itens abaixo, encontre a solução do problema de valor inicial dado. Esboce o gráfico da solução e descreva seu comportamento para valores cada vez maiores de $t$.

a. $y^{\prime\prime}+4y=0, \quad y(0)=0,\ y^{\prime}(0)=1$

b. $y^{\prime\prime}-2y^{\prime}+5y=0, \quad y(\pi/2)=0,\ y^{\prime}(\pi/2)=2$

c. $y^{\prime\prime}+y=0, \quad y(\pi/3)=2,\ y^{\prime}(\pi/3)=-4$

**Exercício 14.** [Seção 3.3, Exerc. 16] Considere o problema de valor inicial

$$
3u^{\prime\prime}-u^{\prime}+2u=0, \qquad u(0)=2,\ u^{\prime}(0)=0.
$$

a. Encontre a solução $u(t)$ deste problema.

b. Para $t>0$, encontre o primeiro instante no qual $|u(t)|=10$.

**Exercício 15.** [Seção 3.3, Exerc. 18] Considere o problema de valor inicial

$$
y^{\prime\prime}+2y^{\prime}+6y=0, \qquad y(0)=2,\ y^{\prime}(0)=\alpha\geq0.
$$

a. Encontre a solução $y(t)$ deste problema.

b. Encontre $\alpha$ tal que $y=0$ quando $t=1$.

c. Encontre o menor valor positivo de $t$, em função de $\alpha$, para o qual $y=0$.

d. Determine o limite da expressão encontrada no item (c) quando $\alpha\to\infty$.

### Equações Não Homogêneas; Coeficientes a Determinar

**Exercício 20.** [Seção 3.5, Exerc. 12,13,15] Em cada um dos itens abaixo, encontre a solução do problema de valor inicial dado.

a. $y^{\prime\prime}+4y=t^2+3e^t, \quad y(0)=0,\ y^{\prime}(0)=2$

b. $y^{\prime\prime}-2y^{\prime}+y=te^t+4, \quad y(0)=1,\ y^{\prime}(0)=1$

c. $y^{\prime\prime}+2y^{\prime}+5y=4e^{-t}\cos(2t), \quad y(0)=1,\ y^{\prime}(0)=0$

**Exercício 21.** [Seção 3.5, Exerc. 16(a),17(a)] Em cada um dos itens abaixo, determine uma forma adequada para $Y(t)$ para usar o método dos coeficientes indeterminados (não é necessário determinar os valores dos coeficientes).

a. $y^{\prime\prime}+3y^{\prime}=2t^4+t^2e^{-3t}+\operatorname{sen}(3t)$

b. $y^{\prime\prime}-5y^{\prime}+6y=e^t\cos(2t)+e^{2t}(3t+4)\operatorname{sen}\,t$

### Vibrações Mecânicas e Elétricas

**Exercício 25.** [Seção 3.7, Exerc. 3] Uma massa de 100 g estica uma mola de 5 cm. Se a massa for colocada em movimento a partir de sua posição de equilíbrio com uma velocidade apontando para baixo de 10 cm/s, e se não houver amortecimento, determine a posição $u$ da massa em qualquer instante $t$. Quando a massa retorna pela primeira vez à sua posição de equilíbrio?

**Exercício 26.** [Seção 3.7, Exerc. 4] Uma massa pesando 3 lb ($\cong1{,}36$ kg) estica uma mola de 3 polegadas ($\cong7{,}6$ cm). Se a massa for empurrada para cima, contraindo a mola de uma polegada, e depois for colocada em movimento com uma velocidade para baixo de 2 pés/s (cerca de 61 cm/s), e se não houver amortecimento, encontre a posição $u$ da massa em qualquer instante $t$. Determine a frequência, o período, a amplitude e a fase do movimento.

**Exercício 27.** [Seção 3.7, Exerc. 20(a),21(a)] Em cada um dos itens abaixo, a posição de determinado sistema mola-massa satisfaz o problema de valor inicial dado. Encontre a solução deste problema de valor inicial.

a. $u^{\prime\prime}+2u=0, \quad u(0)=0,\ u^{\prime}(0)=2$

b. $u^{\prime\prime}+\dfrac14u^{\prime}+2u=0, \quad u(0)=0,\ u^{\prime}(0)=2$

### Vibrações Forçadas

**Exercício 28.** [Seção 3.8, Exerc. 4,5(a,b)] Uma massa de 5 kg estica uma mola de 10 cm. A massa sofre a ação de uma força externa de $10\operatorname{sen}(t/2)$ N (newton) e se move em um meio que amortece o movimento com uma força viscosa de 2 N quando a velocidade da massa é de 4 cm/s. A massa é colocada em movimento a partir de sua posição de equilíbrio com uma velocidade inicial de 3 cm/s.

a. Formule o problema de valor inicial que descreve o movimento da massa.

b. Encontre a solução do problema.

c. Identifique as partes transiente e estado estacionário da solução.

**Exercício 29.** [Seção 3.8, Exerc. 13(a),14(a)] Em cada um dos itens abaixo, encontre a solução $u(t)$, para $\omega\neq1$, do problema de valor inicial dado (o sistema é forçado, mas não amortecido).

a. $u^{\prime\prime}+u=3\cos(\omega t), \quad u(0)=0,\ u^{\prime}(0)=0$

b. $u^{\prime\prime}+u=3\cos(\omega t), \quad u(0)=1,\ u^{\prime}(0)=1$
