# Sistemas Lineares Homogêneos

Vamos agora resolver sistemas lineares homogêneos com coeficientes constantes,

$$
\mathbf{x}' = A\,\mathbf{x},
$$

em que $A$ é uma matriz $n\times n$ constante (no nosso caso, $2\times2$ ou $3\times3$). Assim como fizemos para as equações de segunda ordem e de ordem superior, começamos com um resultado geral que diz **o que** procurar, e depois uma conta que diz **como** encontrar.

## O resultado geral

**Teorema.** Seja $A$ uma matriz $n\times n$ constante.

1. **(Existência e unicidade)** Para qualquer vetor $\mathbf{x}_0$, o PVI $\mathbf{x}'=A\mathbf{x},\ \mathbf{x}(t_0)=\mathbf{x}_0$ tem uma única solução, definida para todo $t$.

2. **(Superposição)** Se $\mathbf{x}^{(1)}(t),\dots,\mathbf{x}^{(k)}(t)$ são soluções, qualquer combinação $c_1\mathbf{x}^{(1)}+\cdots+c_k\mathbf{x}^{(k)}$ também é solução.

3. **(Solução geral)** Se $\mathbf{x}^{(1)},\dots,\mathbf{x}^{(n)}$ são $n$ soluções tais que o determinante
$$
W(t) = \det\big(\,\mathbf{x}^{(1)}(t)\ \ \mathbf{x}^{(2)}(t)\ \cdots\ \mathbf{x}^{(n)}(t)\,\big)
$$
(com as soluções colocadas como colunas) é diferente de zero, então **toda** solução do sistema é da forma
$$
\mathbf{x}(t) = c_1\mathbf{x}^{(1)}(t) + c_2\mathbf{x}^{(2)}(t) + \cdots + c_n\mathbf{x}^{(n)}(t).
$$

4. **(Existência de soluções suficientes)** Sempre é possível encontrar essas $n$ soluções a partir dos autovalores e autovetores de $A$.

O item 3 é o mesmo raciocínio do Wronskiano da segunda ordem. Para ajustar uma condição inicial $\mathbf{x}(t_0)=\mathbf{x}_0$, precisamos resolver o sistema linear algébrico $c_1\mathbf{x}^{(1)}(t_0)+\cdots+c_n\mathbf{x}^{(n)}(t_0)=\mathbf{x}_0$, cuja matriz tem as soluções como colunas. Ele tem solução única para todo $\mathbf{x}_0$ exatamente quando o determinante $W(t_0)$ não é zero — ou seja, quando os vetores $\mathbf{x}^{(1)}(t_0),\dots,\mathbf{x}^{(n)}(t_0)$ são linearmente independentes. (Pode-se mostrar que, se $W$ não se anula em um ponto, não se anula em nenhum.)

Então, para um sistema $2\times2$, a tarefa é encontrar **duas** soluções independentes; para um $3\times3$, **três**.

## De onde vem a solução: $\mathbf{x}=e^{\lambda t}\mathbf{v}$

Para uma equação escalar $x'=ax$, a solução é uma exponencial, $x=ce^{at}$. Para o sistema $\mathbf{x}'=A\mathbf{x}$, é natural tentar algo parecido: uma exponencial multiplicando um vetor constante,

$$
\mathbf{x}(t) = e^{\lambda t}\,\mathbf{v},
$$

com $\lambda$ um número e $\mathbf{v}\neq\mathbf{0}$ um vetor constante, ambos a determinar. Derivando, $\mathbf{x}'=\lambda e^{\lambda t}\mathbf{v}$. Substituindo no sistema:

$$
\lambda e^{\lambda t}\mathbf{v} = A\big(e^{\lambda t}\mathbf{v}\big) = e^{\lambda t}A\mathbf{v}.
$$

Como $e^{\lambda t}\neq0$, podemos cancelar:

$$
A\mathbf{v} = \lambda\mathbf{v}.
$$

Ou seja, $\mathbf{x}=e^{\lambda t}\mathbf{v}$ é solução **exatamente quando $\lambda$ é um autovalor de $A$ e $\mathbf{v}$ é um autovetor correspondente**. É o mesmo papel que a equação característica $ar^2+br+c=0$ fazia na segunda ordem — e, de fato, é a mesma equação quando o sistema vem de uma equação de segunda ordem, como vimos na aula de autovalores.

Então a receita é:

1. encontrar os autovalores, resolvendo $\det(A-\lambda I)=0$;
2. para cada autovalor, encontrar um autovetor, resolvendo $(A-\lambda I)\mathbf{v}=\mathbf{0}$;
3. cada par dá uma solução $e^{\lambda t}\mathbf{v}$; se conseguirmos $n$ delas independentes, a solução geral é a combinação de todas.

Como para a segunda ordem, aparecem três casos, conforme os autovalores sejam **reais e distintos**, **complexos** ou **repetidos**. Cada um terá sua própria seção.

### Um primeiro exemplo

Na aula de autovalores, vimos que $A=\begin{pmatrix} 3 & -1\\ 4 & -2\end{pmatrix}$ tem autovalores $\lambda_1=2$ e $\lambda_2=-1$, com autovetores $\mathbf{v}^{(1)}=\begin{pmatrix} 1\\ 1\end{pmatrix}$ e $\mathbf{v}^{(2)}=\begin{pmatrix} 1\\ 4\end{pmatrix}$. Então

$$
\mathbf{x}^{(1)}(t) = \begin{pmatrix} 1\\ 1\end{pmatrix}e^{2t}, \qquad \mathbf{x}^{(2)}(t) = \begin{pmatrix} 1\\ 4\end{pmatrix}e^{-t}
$$

são soluções de $\mathbf{x}'=A\mathbf{x}$. Elas são independentes:

$$
W(t) = \begin{vmatrix} e^{2t} & e^{-t}\\ e^{2t} & 4e^{-t}\end{vmatrix} = 4e^{t} - e^{t} = 3e^{t} \neq 0.
$$

Logo, pelo teorema, a solução geral é

$$
\mathbf{x}(t) = c_1\begin{pmatrix} 1\\ 1\end{pmatrix}e^{2t} + c_2\begin{pmatrix} 1\\ 4\end{pmatrix}e^{-t}.
$$

## Plano de fase e estabilidade

Para sistemas $2\times2$, vamos sempre desenhar as soluções no **plano de fase** $x_1x_2$, como fizemos com o oscilador na aula de modelagem. O desenho com várias trajetórias é o **retrato de fase** do sistema.

A origem $\mathbf{x}=\mathbf{0}$ é sempre uma solução (constante) de $\mathbf{x}'=A\mathbf{x}$: é o **ponto de equilíbrio** do sistema. Assim como fizemos com a reta de fase das equações autônomas, vamos classificar esse equilíbrio:

* **assintoticamente estável**, se todas as soluções tendem a $\mathbf{0}$ quando $t\to\infty$;
* **estável** (mas não assintoticamente), se as soluções que começam perto de $\mathbf{0}$ ficam perto, sem necessariamente tender a $\mathbf{0}$;
* **instável**, se existem soluções começando tão perto de $\mathbf{0}$ quanto se queira que se afastam dele.

Como as soluções são combinações de termos $e^{\lambda t}$, já dá para adivinhar que quem decide a estabilidade é o **sinal (da parte real) dos autovalores** — exatamente como o sinal das raízes da equação característica decidia o comportamento das vibrações livres. No exemplo acima, como $\lambda_1=2>0$, o termo $e^{2t}$ cresce e a origem é instável. Vamos ver isso caso a caso.
