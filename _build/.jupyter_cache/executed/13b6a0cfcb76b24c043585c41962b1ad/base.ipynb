```python
import sympy as sp

t, c1, c2 = sp.symbols('t c1 c2')
y = sp.exp(t)*(c1*sp.cos(2*t) + c2*sp.sin(2*t))

print("y'' - 2y' + 5y =", sp.simplify(sp.diff(y, t, 2) - 2*sp.diff(y, t) + 5*y))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def desenha_vetor(z, ax, cor, rotulo):
    ax.annotate('', xy=(z.real, z.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle='-|>', color=cor, lw=2.5))
    ax.annotate(rotulo, xy=(z.real, z.imag),
                xytext=(z.real*1.2 + 0.15, z.imag*1.2 + 0.15), fontsize=12)

fig, ax = plt.subplots(figsize=(6, 6))

z = 2 + 0j
zs = [z, 1j*z, 1j**2*z, 1j**3*z]
rotulos = ['$2$', '$2i$', '$-2$', '$-2i$']
cores = ['b', 'g', 'r', 'purple']

for zz, rot, cor in zip(zs, rotulos, cores):
    desenha_vetor(zz, ax, cor, rot)

lim = 3
ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim)
ax.axhline(0, color='0.7'); ax.axvline(0, color='0.7')
ax.set_aspect('equal')
ax.set_xlabel('parte real'); ax.set_ylabel('parte imaginária')
ax.set_title(r"Multiplicar por $i$ gira $90°$")
ax.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

z = 3 + 1j
zs = [z, 1j*z, 2j*z]
rotulos = ['$z=3+i$', '$iz$', '$2iz$']
cores = ['b', 'g', 'r']

for zz, rot, cor in zip(zs, rotulos, cores):
    desenha_vetor(zz, ax, cor, rot)

lim = 8
ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim)
ax.axhline(0, color='0.7'); ax.axvline(0, color='0.7')
ax.set_aspect('equal')
ax.set_xlabel('parte real'); ax.set_ylabel('parte imaginária')
ax.set_title(r"Multiplicar $z$ por $i$ (gira) e por $2i$ (gira e dobra)")
ax.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

```python
theta = sp.symbols('theta', real=True)

serie_exp = sp.series(sp.exp(sp.I*theta), theta, 0, 8).removeO()
serie_euler = sp.series(sp.cos(theta) + sp.I*sp.sin(theta), theta, 0, 8).removeO()

print("Série de e^(i*theta)         :", serie_exp)
print("Série de cos(theta)+i*sen(theta):", serie_euler)
print("Diferença                    :", sp.simplify(serie_exp - serie_euler))
```
