```python
import sympy as sp

t, c1, c2 = sp.symbols('t c1 c2')

def residuo(y, a, b, c, F):
    return sp.simplify(a*sp.diff(y, t, 2) + b*sp.diff(y, t) + c*y - F)

y1 = c1*sp.exp(t) + c2*sp.exp(2*t) + sp.Rational(1, 6)*sp.exp(4*t)
print("Resíduo do Exemplo 1:", residuo(y1, 1, -3, 2, sp.exp(4*t)))
```

```python
y2 = c1*sp.exp(-2*t) + c2*sp.exp(t) - sp.Rational(3, 10)*sp.cos(t) + sp.Rational(1, 10)*sp.sin(t)
print("Resíduo do Exemplo 2:", residuo(y2, 1, 1, -2, sp.cos(t)))
```

```python
y3 = c1*sp.exp(2*t) + c2*sp.exp(-t) - t/2 + sp.Rational(1, 4)
print("Resíduo do Exemplo 3:", residuo(y3, 1, -1, -2, t))
```

```python
y4 = c1*sp.cos(2*t) + c2*sp.sin(2*t) + sp.exp(t)*(sp.Rational(1, 5)*sp.cos(t) + sp.Rational(1, 10)*sp.sin(t))
print("Resíduo do Exemplo 4:", residuo(y4, 1, 0, 4, sp.exp(t)*sp.cos(t)))
```

```python
y5 = c1*sp.exp(t) + c2*sp.exp(2*t) + t*sp.exp(2*t)
print("Resíduo do Exemplo 5:", residuo(y5, 1, -3, 2, sp.exp(2*t)))
```

```python
y6 = (c1 + c2*t)*sp.exp(2*t) + sp.Rational(1, 2)*t**2*sp.exp(2*t)
print("Resíduo do Exemplo 6:", residuo(y6, 1, -4, 4, sp.exp(2*t)))
```

```python
y7 = c1*sp.cos(2*t) + c2*sp.sin(2*t) + (t/4)*sp.sin(2*t)
print("Resíduo do Exemplo 7:", residuo(y7, 1, 0, 4, sp.cos(2*t)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

c1_, c2_ = 0, 0  # apenas para ilustrar y_p isoladamente (não é PVI da aula)
t_vals = np.linspace(0, 40, 400)
y_vals = c1_*np.cos(2*t_vals) + c2_*np.sin(2*t_vals) + (t_vals/4)*np.sin(2*t_vals)

plt.figure(figsize=(9, 4.5))
plt.plot(t_vals, y_vals, 'b', linewidth=1.5)
plt.plot(t_vals, t_vals/4, 'r--', linewidth=1.5, label=r'envelope $\pm t/4$')
plt.plot(t_vals, -t_vals/4, 'r--', linewidth=1.5)
plt.title(r"Ressonância: $y_p(t)=\frac{t}{4}\,\mathrm{sen}(2t)$ — amplitude crescendo linearmente")
plt.xlabel("$t$"); plt.ylabel("$y$")
plt.legend()
plt.grid(True)
plt.show()
```
