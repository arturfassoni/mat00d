```python
import numpy as np
import sympy as sp

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 0], [-1, 2]])
x = np.array([3, -1])

print("A + B =\n", A + B)
print("3A =\n", 3*A)
print("A^T =\n", A.T)
print("AB =\n", A @ B)
print("BA =\n", B @ A)
print("Ax =", A @ x)
```

```python
A = sp.Matrix([[1, -1, -1],
               [3, -1,  2],
               [2,  2,  3]])

print("det A =", A.det())
Ainv = A.inv()
display(Ainv)
display(A * Ainv)
```
