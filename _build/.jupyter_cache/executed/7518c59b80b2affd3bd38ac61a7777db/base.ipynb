```python
import sympy as sp

t = sp.symbols('t')
y = (1 + 2*t)*sp.exp(3*t)

print("y'' - 6y' + 9y =", sp.simplify(sp.diff(y, t, 2) - 6*sp.diff(y, t) + 9*y))
print("y(0)  =", y.subs(t, 0))
print("y'(0) =", sp.diff(y, t).subs(t, 0))
```

```python
import numpy as np
import matplotlib.pyplot as plt

t_vals = np.linspace(-1, 1, 200)
y_vals = (1 + 2*t_vals)*np.exp(3*t_vals)

plt.figure(figsize=(8, 5))
plt.plot(t_vals, y_vals, 'b', linewidth=2.5)
plt.plot(0, 1, 'ro', markersize=7, label='$(0,\\,1)$')
plt.axhline(0, color='0.7', linewidth=0.8)
plt.title(r"Solução de $y''-6y'+9y=0,\ y(0)=1,\ y'(0)=5$")
plt.xlabel("$t$"); plt.ylabel("$y$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
y2 = (3 + 7*t)*sp.exp(-2*t)

print("y'' + 4y' + 4y =", sp.simplify(sp.diff(y2, t, 2) + 4*sp.diff(y2, t) + 4*y2))
print("y(0)  =", y2.subs(t, 0))
print("y'(0) =", sp.diff(y2, t).subs(t, 0))

t_vals = np.linspace(0, 3, 200)
y_vals = (3 + 7*t_vals)*np.exp(-2*t_vals)

plt.figure(figsize=(8, 5))
plt.plot(t_vals, y_vals, 'b', linewidth=2.5)
plt.axhline(0, color='0.7', linewidth=0.8)
plt.title(r"Solução de $y''+4y'+4y=0,\ y(0)=3,\ y'(0)=1$")
plt.xlabel("$t$"); plt.ylabel("$y$")
plt.grid(True)
plt.show()
```

```python
a, b, c, r, t_s = sp.symbols('a b c r t')
u = sp.Function('u')

y_geral = u(t_s)*sp.exp(r*t_s)
expr = sp.expand((a*sp.diff(y_geral, t_s, 2) + b*sp.diff(y_geral, t_s) + c*y_geral) / sp.exp(r*t_s))
expr = sp.collect(expr, [u(t_s).diff(t_s, 2), u(t_s).diff(t_s), u(t_s)])
print("Forma geral, antes de usar a condição de raiz repetida:")
print(expr)

# condição de raiz repetida: r = -b/(2a)  <=>  b = -2ar ;  e a*r^2+b*r+c=0 <=> c = a*r^2
expr_repetida = expr.subs({b: -2*a*r, c: a*r**2})
print("\nUsando a condição de raiz repetida (b=-2ar, c=ar^2):")
print(sp.simplify(expr_repetida))
```
