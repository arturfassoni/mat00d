# Equações Não Homogêneas de Ordem Superior

Para a equação não homogênea

$$
a_n y^{(n)} + \cdots + a_1 y' + a_0 y = F(t),
$$

vale exatamente o mesmo resultado que vimos para a segunda ordem, com a mesma justificativa (a diferença entre duas soluções da não homogênea resolve a homogênea):

**Resultado.** A solução geral é $y=y_h+y_p$, em que $y_h=c_1y_1+\cdots+c_ny_n$ é a solução geral da homogênea associada e $y_p$ é **uma** solução particular qualquer.

E o método dos coeficientes a determinar funciona do mesmo jeito: tentamos $y_p$ com a "cara" de $F(t)$, com uma correção quando a tentativa colide com soluções da homogênea:

**Regra.** Se $F(t)=P_m(t)\,e^{\alpha t}\cos\beta t$ ou $P_m(t)\,e^{\alpha t}\,\text{sen}\,\beta t$ (com $P_m$ um polinômio de grau $m$), tente

$$
y_p = t^s e^{\alpha t}\Big[\big(A_0+A_1t+\cdots+A_mt^m\big)\cos\beta t + \big(B_0+B_1t+\cdots+B_mt^m\big)\,\text{sen}\,\beta t\Big],
$$

em que $s$ é a **multiplicidade de $\alpha+i\beta$ como raiz de $Z(r)$** ($s=0$ se não for raiz). Os casos $\beta=0$ (exponenciais e polinômios) e $\alpha=0$ (senos e cossenos) estão incluídos. Se $F(t)$ for uma soma de termos desse tipo, tratamos cada termo separadamente e somamos as soluções particulares.

A única novidade em relação à segunda ordem: como as raízes podem ter multiplicidade maior, $s$ pode ser $3$, $4$, ... e não apenas $0$, $1$ ou $2$.

### Exemplo 1: sem ressonância

Encontre a solução geral de

$$
y''' - 6y'' + 11y' - 6y = 12e^{4t}.
$$

A homogênea associada foi resolvida na seção anterior: $Z(r)=(r-1)(r-2)(r-3)$ e $y_h=c_1e^{t}+c_2e^{2t}+c_3e^{3t}$.

Como $4$ não é raiz de $Z(r)$, não há colisão ($s=0$) e tentamos $y_p=Ae^{4t}$. Aqui vale um atalho útil: para qualquer $\alpha$,

$$
a_n\big(e^{\alpha t}\big)^{(n)}+\cdots+a_0e^{\alpha t} = Z(\alpha)\,e^{\alpha t},
$$

pois cada derivada só multiplica por $\alpha$. Então, substituindo $y_p=Ae^{4t}$:

$$
A\,Z(4)\,e^{4t} = 12e^{4t}, \qquad Z(4) = 64-96+44-6 = 6 \quad\Longrightarrow\quad A=2.
$$

A solução geral é

$$
y = c_1e^{t}+c_2e^{2t}+c_3e^{3t}+2e^{4t}.
$$

(O atalho só funciona porque $Z(4)\neq0$; quando $\alpha$ é raiz, temos ressonância — próximo exemplo.)

### Exemplo 2: ressonância tripla

Encontre a solução geral de

$$
y''' - 3y'' + 3y' - y = 4e^{t}.
$$

A homogênea associada também já foi resolvida: $Z(r)=(r-1)^3$, e $y_h=(c_1+c_2t+c_3t^2)e^{t}$.

A tentativa natural seria $Ae^{t}$ — mas $e^{t}$, $te^{t}$ e $t^2e^{t}$ são **todas** soluções da homogênea. A regra diz: $\alpha=1$ é raiz de multiplicidade $s=3$, então

$$
y_p = At^3e^{t}.
$$

Derivando três vezes, substituindo e simplificando (os termos com $t^3$, $t^2$ e $t$ se cancelam — vale a pena conferir), sobra

$$
6Ae^{t} = 4e^{t} \quad\Longrightarrow\quad A=\tfrac23.
$$

A solução geral é

$$
y = (c_1+c_2t+c_3t^2)\,e^{t} + \tfrac23t^3e^{t}.
$$

### Exemplo 3: superposição, com e sem ressonância

Encontre uma solução particular de

$$
y''' - 4y' = t + 3\cos t + e^{-2t}.
$$

**Homogênea associada:** $Z(r)=r^3-4r=r(r-2)(r+2)$, com raízes $0$, $2$, $-2$, e $y_h=c_1+c_2e^{2t}+c_3e^{-2t}$.

O lado direito tem três termos de tipos diferentes; tratamos cada um separadamente:

* **$F_1=t$** (polinômio, $\alpha=\beta=0$): a tentativa seria $A_0t+A_1$, mas a constante $A_1$ é solução da homogênea ($r=0$ é raiz simples, $s=1$). Logo $Y_1=t(A_0t+A_1)$.
* **$F_2=3\cos t$** ($\alpha=0,\ \beta=1$): $\pm i$ não é raiz, $s=0$. Logo $Y_2=B\cos t+C\,\text{sen}\,t$.
* **$F_3=e^{-2t}$** ($\alpha=-2$): $-2$ é raiz simples, $s=1$. Logo $Y_3=Ete^{-2t}$.

Substituindo cada tentativa na equação com o termo correspondente:

* $Y_1$: $Y_1'''-4Y_1'=-4(2A_0t+A_1)=t \Rightarrow A_0=-\tfrac18,\ A_1=0$.
* $Y_2$: $Y_2'''-4Y_2'=5B\,\text{sen}\,t-5C\cos t=3\cos t \Rightarrow B=0,\ C=-\tfrac35$.
* $Y_3$: após as derivadas, $8Ee^{-2t}=e^{-2t} \Rightarrow E=\tfrac18$.

Somando:

$$
y_p = -\frac{t^2}{8} - \frac35\,\text{sen}\,t + \frac18\,te^{-2t}.
$$

Repare no primeiro termo: o lado direito tinha só $t$ (grau $1$), mas a solução particular tem $t^2$. É a ressonância com a raiz $r=0$ — a mesma ideia de "multiplicar por $t$", aplicada a um polinômio.

## Resumo

Para resolver $a_ny^{(n)}+\cdots+a_0y=F(t)$ com coeficientes constantes:

1. Encontre as raízes de $Z(r)=a_nr^n+\cdots+a_0$ (é aqui que mora a maior parte do trabalho) e monte $y_h$ pela tabela da seção anterior.
2. Para cada termo de $F(t)$, monte a tentativa de $y_p$ pela regra acima, com $t^s$ e $s$ igual à multiplicidade da raiz correspondente.
3. Substitua, resolva para os coeficientes, e some: $y=y_h+y_p$.
4. Se houver condições iniciais, imponha as $n$ condições na solução geral (não em $y_h$!) e resolva o sistema $n\times n$.

Quando $F(t)$ não tem uma das formas da tabela, o método de variação dos parâmetros também se estende para ordem $n$, mas exige resolver sistemas $n\times n$ e calcular integrais — e, como na segunda ordem, não vamos abordá-lo neste curso.
