## Exercícios

Exercícios retirados da 9a Edição do livro _Equações Diferenciais e Problemas de Valores de Contorno_ de Boyce e DiPrima.

### Equações Homogêneas

**Exercício 1.** [Seção 4.2, Exerc. 8,14,16] Em cada um dos itens abaixo, encontre a solução geral da equação diferencial dada.

a. $y^{\prime\prime\prime}-y^{\prime\prime}-y^{\prime}+y=0$

b. $y^{(5)}-3y^{(4)}+3y^{\prime\prime\prime}-3y^{\prime\prime}+2y^{\prime}=0$

c. $y^{(4)}+2y^{\prime\prime}+y=0$

**Exercício 2.** [Seção 4.2, Exerc. 22,25] Em cada um dos itens abaixo, encontre a solução do problema de valor inicial dado e faça seu gráfico. Como a solução se comporta quando $t\to\infty$?

a. $y^{(4)}-4y^{\prime\prime\prime}+4y^{\prime\prime}=0; \quad y(1)=-1,\ y^{\prime}(1)=2,\ y^{\prime\prime}(1)=0,\ y^{\prime\prime\prime}(1)=0$

b. $6y^{\prime\prime\prime}+5y^{\prime\prime}+y^{\prime}=0; \quad y(0)=-2,\ y^{\prime}(0)=2,\ y^{\prime\prime}(0)=0$

### Equações Não Homogêneas

**Exercício 3.** [Seção 4.3, Exerc. 2] Determine a solução geral da equação diferencial dada:

$$
y^{(4)}-y=3t+\cos t.
$$

**Exercício 4.** [Seção 4.3, Exerc. 7] Encontre a solução do problema de valor inicial dado. Depois faça um gráfico da solução.

$$
y^{\prime\prime\prime}+4y^{\prime}=t; \qquad y(0)=y^{\prime}(0)=0,\ y^{\prime\prime}(0)=1.
$$

**Exercício 5.** [Seção 4.3, Exerc. 10,13] Em cada um dos itens abaixo, determine uma forma adequada para $Y(t)$ se for utilizado o método dos coeficientes indeterminados. Não calcule as constantes.

a. $y^{\prime\prime\prime}-2y^{\prime\prime}+y^{\prime}=t^3+2e^t$

b. $y^{(4)}+2y^{\prime\prime\prime}+2y^{\prime\prime}=3e^t+2te^{-t}+e^{-t}\operatorname{sen} t$
