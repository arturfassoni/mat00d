## Exercícios

Exercícios retirados da 9a Edição do livro _Equações Diferenciais e Problemas de Valores de Contorno_ de Boyce e DiPrima.

### Modelagem com Sistemas

**Exercício 1.** [Seção 7.1, Exerc. 3] Transforme a equação dada em um sistema de equações de primeira ordem:

$$
u^{(4)}-u=0.
$$

**Exercício 2.** [Seção 7.1, Exerc. 4] Transforme o problema de valor inicial dado em um problema de valor inicial para duas equações de primeira ordem:

$$
u^{\prime\prime}+0{,}25u^{\prime}+4u=2\cos(3t), \qquad u(0)=1,\ u^{\prime}(0)=-2.
$$

### Matrizes

**Exercício 3.** [Seção 7.2, Exerc. 17,18] Em cada um dos itens abaixo, verifique que o vetor dado satisfaz a equação diferencial dada.

a. $\mathbf{x}^{\prime}=\begin{pmatrix} 2 & -1\\ 3 & -2\end{pmatrix}\mathbf{x}+\begin{pmatrix} 1\\ -1\end{pmatrix}e^t, \qquad \mathbf{x}=\begin{pmatrix} (1+2t)e^t\\ 2te^t\end{pmatrix}=\begin{pmatrix} 1\\ 0\end{pmatrix}e^t+2\begin{pmatrix} 1\\ 1\end{pmatrix}te^t$

b. $\mathbf{x}^{\prime}=\begin{pmatrix} 1 & 1 & 1\\ 2 & 1 & -1\\ 0 & -1 & 1\end{pmatrix}\mathbf{x}, \qquad \mathbf{x}=\begin{pmatrix} 6e^{-t}\\ -8e^{-t}+2e^{2t}\\ -4e^{-t}-2e^{2t}\end{pmatrix}=\begin{pmatrix} 6\\ -8\\ -4\end{pmatrix}e^{-t}+2\begin{pmatrix} 0\\ 1\\ -1\end{pmatrix}e^{2t}$

**Exercício 4.** [Seção 7.2, Exerc. 19] Verifique se a matriz dada satisfaz a equação diferencial dada:

$$
\Psi^{\prime}=\begin{pmatrix} 1 & 1\\ 4 & -2\end{pmatrix}\Psi, \qquad \Psi(t)=\begin{pmatrix} e^{-3t} & e^{2t}\\ -4e^{-3t} & e^{2t}\end{pmatrix}.
$$

### Autovalores e Autovetores

**Exercício 5.** [Seção 7.3, Exerc. 14,16,17] Em cada um dos itens abaixo, encontre todos os autovalores e autovetores da matriz dada.

a. $\begin{pmatrix} 5 & -1\\ 3 & 1\end{pmatrix}$

b. $\begin{pmatrix} -2 & 1\\ 1 & -2\end{pmatrix}$

c. $\begin{pmatrix} 1 & \sqrt3\\ \sqrt3 & -1\end{pmatrix}$
