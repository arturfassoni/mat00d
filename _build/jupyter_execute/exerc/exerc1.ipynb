## Exercícios P1

Exercícios retirados da 9a Edição do livro _Equações Diferenciais e Problemas de Valores de Contorno_ de Boyce e DiPrima.

**Exercício 1.** [Seção 1.1, Exerc. 1-4] Em cada um dos itens abaixo, desenhe um campo de direçöes para a equação diferencial dada. Baseado no campo de direções, determine o comportamento de $y$ quando $t \rightarrow \infty$. Se esse comportamento depender do valor inicial de $y$ em $t=0$, descreva essa dependência.

a. $y'=3-2 y$

b. $y'=2 y-3$

c. $y'=-1-2 y$

d. $y'=1+2 y$


**Exercício 2.** [Seção 1.3, Exerc. 5-7] Em cada um dos itens abaixo, verifique que cada função dada é solução da equação diferencial.

a. $\quad y^{\prime \prime}-y=0 ; \quad y_1(t)=e^t, \quad y_2(t)=\cosh t$

b. $\quad y^{\prime \prime}+2 y^{\prime}-3 y=0 ; \quad y_1(t)=e^{-3 t}, \quad y_2(t)=e^t$

c. $\quad t y^{\prime}-y=t^2 ; \quad y=3 t+t^2$

**Exercício 3.** [Seção 1.3, Exerc. 1-4] Em cada um dos itens abaixo, determine a ordem da equação diferencial dada e diga se ela é linear ou não linear.

a. $t^2 \frac{d^2 y}{d t^2}+t \frac{d y}{d t}+2 y=\operatorname{sen} t$

b. $\left(1+y^2\right) \frac{d^2 y}{d t^2}+t \frac{d y}{d t}+y=e^t$

c. $\frac{d^4 y}{d t^4}+\frac{d^3 y}{d t^3}+\frac{d^2 y}{d t^2}+\frac{d y}{d t}+y=1$

d. $\frac{d^2 y}{d t^2}+\operatorname{sen}(t+y)=\operatorname{sen} t$

**Exercício 4.** [Seção 1.2, Exerc. 7] Uma população de ratos do campo satisfaz a equação diferencial (veja a seção Modelagem):

$$
\frac{d y}{d t}=\frac{p}{2}-450 .
$$

a. Encontre o instante em que a população é extinta se $p(0)=850$.

b. Encontre o instante de extinção se $p(0)=p_0$, em que $0<p_0<900$.

c. Encontre a população inicial $p_0$ se a população se torna extinta em um ano.

**Exercício 5.** [Seção 1.2, Exerc. 9] Considere um objeto de massa 10 kg em queda seguindo o modelo de queda livre (seção Modelagem), mas suponha agora que o coeficiente de resistência do ar é proporcional ao quadrado da velocidade.

a. Se a velocidade limite for $49 \mathrm{~m} / \mathrm{s}$, mostre que a equação de movimento pode ser escrita como

$$
\frac{d v}{d t}=\frac{1}{245}\left(49^2-v^2\right) .
$$

b. Se $v(0)=0$, encontre uma expressão para $v(t)$ em qualquer instante.

c. Faça o gráfico da solução encontrada no item b e da solução (26) do Exemplo 2 no mesmo conjunto de eixos.

d. Baseado nos gráficos encontrados no item c, compare o efeito de um coeficiente de resistência do ar quadrático com um linear.

e. Encontre a distância $x(t)$ percorrida pelo objeto até o instante $t$.

f. Encontre o tempo $T$ que demora para o objeto cair 300 m.

**Exercício 6.** [Seção 1.2, Exerc. 13] Considere um circuito elétrico contendo um capacitor, um resistor e uma bateria em série. A carga $Q(t)$ no capacitor satisfaz a equação

$$
R \frac{d Q}{d t}+\frac{Q}{C}=V,
$$

em que $R$ é a resistência, $C$ é a capacitância e $V$ é a voltagem constante fornecida pela bateria. (Você não precisa saber de circuitos elétricos aqui, mas apenas usar o modelo acima para responder os itens a seguir).

a. Se $Q(0)=0$, encontre $Q(t)$ em qualquer instante $t$ e esboce o gráfico de $Q$ em função de $t$.

b.Encontre o valor limite $Q_L$ para onde $Q(t)$ tende após um longo período de tempo.

c. Suponha que $Q\left(t_1\right)=Q_L$ e que, no instante $t=t_1$, a bateria é removida e o circuito é fechado novamente. Encontre $Q(t)$ para $t>t_1$ e esboce seu gráfico.
